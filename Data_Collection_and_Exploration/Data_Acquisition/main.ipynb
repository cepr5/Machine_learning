**ВАЖНО!** *Что бы не путаться с датами, всегда стоит отдавать предпочтение формату ISO 8601: YYYY-MM-DD или YYYY-MM-DD HH:MM:SS. А любые расчёты можно проводить, переводя этот формат в одно число. Таким образом легко посчитать следующий день. Допустим если наше число в секундах, то следующий день мы получим прибавив 24ч * 60 мин * 60 сек = +86400 сек*
