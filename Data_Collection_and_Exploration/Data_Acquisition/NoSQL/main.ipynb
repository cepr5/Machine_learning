# NoSQL

Тема довольно обширная и у каждой базы свои правила, в отличии от баз на основе SQL. Но есть как минимум 4 самых основных типов NoSQL:
| Тип | Что хранит | Как работает | Примеры |
|---|---|---|---|
| **Document** | Документы (JSON/BSON) с гибкой схемой | Каждая запись — отдельный документ (обычно JSON) с произвольной структурой. Поля могут быть вложенными, у разных документов — разный набор полей. Поиск идёт по содержимому документа. | MongoDB, CouchDB, Firestore |
| **Key-Value** | Пары «ключ → значение» | Простейшая модель: по уникальному ключу достаётся значение (строка, число, объект). Нет запросов по содержимому — только «дай значение по ключу». Максимально быстро. | Redis, Memcached, DynamoDB |
| **Column (wide-column)** | Данные по столбцам, огромные объёмы | Данные хранятся не построчно, а по семействам столбцов. Строки могут иметь разный набор столбцов. Оптимизировано под огромные объёмы и распределённость по многим серверам. | Cassandra, HBase, Bigtable |
| **Graph** | Узлы и связи | Данные — это узлы (объекты) и рёбра (связи между ними). Запросы — это обход графа: «найди всех друзей друзей», «кратчайший путь». Оптимизировано под анализ связей. | Neo4j, ArangoDB, Neptune |
