# CSV

Основной способ чтения через pandas метод read_csv(), вот основные параметры:

| Параметр | Что делает | Пример |
|---|---|---|
| `sep` | Разделитель колонок | `pd.read_csv("data.txt", sep="\t")` — табуляция вместо запятой |
| `encoding` | Кодировка файла | `pd.read_csv("data.csv", encoding="utf-8-sig")` — если файл с BOM (часто из Excel) |
| `dtype` | Явно задать типы колонок | `pd.read_csv("data.csv", dtype={"inn": str, "phone": str, "zip": str})` — чтобы ведущие нули не потерялись |
| `parse_dates` | Сразу распарсить даты | `pd.read_csv("data.csv", parse_dates=[["date", "time"]])` — склеить две колонки в один datetime |
| `usecols` | Читать только нужные колонки | `pd.read_csv("data.csv", usecols=["id", "date", "amount"])` — экономия памяти на больших файлах |
| `nrows` | Прочитать первые N строк | `pd.read_csv("data.csv", nrows=1000)` — быстрый осмотр перед полной загрузкой |
| `chunksize` | Читать файл частями (итератор DataFrame) | `for chunk in pd.read_csv("big.csv", chunksize=100000): process(chunk)` — обработка файла, который не влезает в память |
| `na_values` | Какие значения считать пропусками | `pd.read_csv("data.csv", na_values=["NA", "-", "", "нет данных", -999])` — несколько маркеров сразу |
| `skiprows` / `header` | Пропустить мусор сверху / указать строку заголовка | `pd.read_csv("data.csv", skiprows=3, header=0)` — пропустить 3 строки отчёта, заголовок в 4-й |
| `decimal` | Десятичный разделитель | `pd.read_csv("data.csv", sep=";", decimal=",", thousands=".")` — европейский формат `1.234,56` |
| `thousands` | Разделитель тысяч | `pd.read_csv("data.csv", thousands=" ")` — числа вида `1 234 567` |

*Важная ловушка! Если есть какие-нибудь, например, серийные номера, то у них могут быть впереди нули "000145", поэтому такие колонки важно приводить к типу str!*
