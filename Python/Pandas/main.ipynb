## Широкоформатный формат в длинноформатный

```python
no2_pivoted = no2.pivot(columns="location", values="value").reset_index()   # Удалили индексы даты (даты стали столбцом) и теперь там просто перечисление от 0 до n
no2_pivoted.head()
```

Тоже довольно сложная операция по переделыванию всего DataFrame:
- id_vars - Столбцы, которые остаются как есть (идентификаторы)
- value_vars - Столбцы, которые нужно расплавить в длинный формат (по умолчанию все кроме id_vars)
- var_name - Имя нового столбца, куда попадут названия расплавленных столбцов
- value_name - Имя нового столбца, куда попадут значения
- col_level - Для MultiIndex-столбцов — какой уровень использовать
- ignore_index - True (по умолчанию) — сбросить старый индекс

*Тут довольно неудачный пример, суть в том, что мы берём столбец id_vars и для каждого другого столбца value_vars создаём отдельную строку. То есть примерно вот так (тут id_vars=name):*

Было:
| name  | math | phys | chem |
|-------|------|------|------|
| Аня   | 5    | 4    | 3    |
| Борис | 4    | 5    | 5    |

Стало:
| name  | subject | grade |
|-------|---------|-------|
| Аня   | math    | 5     |
| Аня   | phys    | 4     |
| Аня   | chem    | 3     |
| Борис | math    | 4     |
| Борис | phys    | 5     |
| Борис | chem    | 5     |

```python
no_2 = no2_pivoted.melt(id_vars="date.utc")
no_2.head()
```

```python
no_2 = no2_pivoted.melt(
    id_vars="date.utc",
    value_vars=["BETR801", "FR04014", "London Westminster"],
    value_name="NO_2",
    var_name="id_location",
)


no_2.head()
```

## Объединение таблиц с использованием общего идентификатора

Самый простой вариант. Берём те варианты, где в обеих таблицах есть одинаковый индекс:

*`left.merge(right...` и `pd.merge(left,right...` одно и тоже*

*`left.merge(right, on=["id", "date"])` - может быть несколько ключей*

```python
left = pd.DataFrame({
    "id":    [1, 2, 3, 4],
    "name":  ["Аня", "Борис", "Вика", "Гена"],
})

right = pd.DataFrame({
    "id":    [3, 4, 5, 6],
    "grade": [5, 4, 3, 5],
})

print(left.merge(right, on="id"))
```

Параметр `how` — тип объединения:
- "inner" - Только совпадающие ключи (по умолчанию)
- "left" - Все строки из left + совпадения из right
- "right" - Все строки из right + совпадения из left
- "outer" - Все строки из обеих таблиц
- "cross" - Декартово произведение

```python
left.merge(right, on="id", how="left")
```

```python
left.merge(right, on="id", how="right")
```

```python
left.merge(right, on="id", how="outer")
```

```python
left.merge(right, how="cross")
```

Параметры left_on / right_on, когда название ключа у таблиц разное:

```python
orders = pd.DataFrame({
    "order_id":  [1, 2, 3],
    "user_id":   [10, 20, 30],
})

users = pd.DataFrame({
    "id":    [10, 20, 40],
    "name":  ["Аня", "Борис", "Вика"],
})

orders.merge(users, left_on="user_id", right_on="id")
```

Параметры left_index / right_index, когда хотим объдинение по индексам:

```python
left.merge(right, left_index=True, right_index=True)   # по двум индексам
```

```python
left.merge(right, left_index=True, right_on="id")   # смешанны вариант
```

Параметр suffixes - если в обеих таблицах есть столбцы с одинаковыми именами (но не ключ), pandas добавит суффиксы:

```python
left = pd.DataFrame({"id": [1], "value": [10]})
right = pd.DataFrame({"id": [1], "value": [20]})

left.merge(right, on="id")
```

```python
left.merge(right, on="id", suffixes=("_left", "_right"))
```

Параметр sort - по умолчанию False — порядок ключей не сортируется. С `sort=True` результат сортируется по ключам

Параметр indicator - добавляет столбец _merge, показывающий, откуда пришла строка:

```python
left.merge(right, on="id", how="outer", indicator=True)
```

*можно дать своё название - `left.merge(right, on="id", how="outer", indicator="source")`* 

Параметр validate - проверяет, что объединение соответствует ожидаемой структуре. Если нет — падает с ошибкой. То есть ключи желательно должны быть уникальны!:
| Значение | Проверяет |
|-|-|
| "1:1" | Ключи уникальны в обеих таблицах |
| "1:m" | Ключи уникальны в left |
| "m:1" | Ключи уникальны в right |
| "m:m" | Ничего не проверяет (по умолчанию) |

Параметр copy - `copy=True` (по умолчанию)
