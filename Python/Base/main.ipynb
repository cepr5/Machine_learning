````python
try:
    # код, который может вызвать исключение
    x = int("abc")
except ValueError as e:
    # обработка исключения
    print("Ошибка преобразования:", e)
except KeyError:
    pass
else:
    # выполняется, если исключения не было
    print("Всё хорошо")
finally:
    # выполняется всегда
    print("Завершение")
````

Краткая иерархия:


Таблица основных исключений:

| Исключение | Когда возникает |
|---|---|
| `Exception` | Базовый класс для обычных ошибок. Используется в `except Exception` для общей обработки. |
| `ValueError` | Аргумент правильного типа, но неподходящего значения. Например, `int("abc")`. |
| `TypeError` | Операция применена к объекту неподходящего типа. Например, `1 + "2"`. |
| `IndexError` | Индекс вне диапазона последовательности. Например, `[1, 2][5]`. |
| `KeyError` | Ключ отсутствует в словаре. Например, `{"a": 1}["b"]`. |
| `AttributeError` | У объекта нет такого атрибута или метода. Например, `"text".unknown()`. |
| `NameError` | Имя не найдено в области видимости. Например, обращение к необъявленной переменной. |
| `ZeroDivisionError` | Деление на ноль. |
| `FileNotFoundError` | Файл или каталог не найден. |
| `ImportError` / `ModuleNotFoundError` | Не удалось импортировать модуль или имя из модуля. |
| `StopIteration` | Итератор закончился. Обычно возникает внутри `next()`. |
| `RuntimeError` | Общая ошибка времени выполнения, когда нет более подходящего исключения. |
| `NotImplementedError` | Абстрактный метод не реализован в наследнике. |
| `RecursionError` | Превышена максимальная глубина рекурсии. |
| `AssertionError` | Условие в `assert` оказалось ложным. |
| `OSError` | Ошибки операционной системы, ввода-вывода, файлов, сети. |
| `PermissionError` | Недостаточно прав доступа к файлу или ресурсу. |
| `TimeoutError` | Истекло время ожидания системной операции. |
| `UnicodeDecodeError` | Не удалось декодировать байты в строку. |
| `UnicodeEncodeError` | Не удалось закодировать строку в байты. |
| `KeyboardInterrupt` | Пользователь прервал программу, обычно Ctrl+C. Наследуется от `BaseException`. |
| `SystemExit` | Программа завершается через `sys.exit()` или `exit()`. Наследуется от `BaseException`. |

## База ООП в Python

Основная суть в трёх китах: Наследование, Инкапсуляция и Полиморфизм

````python
class User:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        return f"Привет, я {self.name}"


u = User("Иван", 25)
print(u.greet())
````

**Наследование и super():**

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "..."

class Dog(Animal):
    def speak(self):
        return f"{self.name}: Гав"
````

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary


class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary)   # вызываем родительский __init__
        self.department = department


m = Manager("Анна", 100000, "IT")
print(m.name, m.salary, m.department)
````

**Композиция:**

````python
class Engine:
    def start(self):
        return "Двигатель запущен"


class Car:
    def __init__(self):
        self.engine = Engine()   # композиция

    def start(self):
        return f"Машина: {self.engine.start()}"


car = Car()
print(car.start())  # Машина: Двигатель запущен
````
