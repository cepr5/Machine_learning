Однако если два вектора в линейной комбинации лежат на одной прямой, то мы получаем не все векторы в пространстве, а только те, что тоже лежат на этой прямой (что подводить нас к линейной зависимости)

В линейной алгебре мы имеем дело только с линейными преобразованиями, что означает, новое пространство должно быть: 1. Из той же точки. 2. Линии должны оставаться прямыми. 3. Они должны быть равноудалены и параллельны:

Композиция, это когда мы применяем сразу несколько преобразований над вектором. И что бы две матрицы преобразований стали композицией, достаточно провести матричное умножение (через @, строки на столбцы). Не забывая что порядок важен! Ещё условно это можно понимать как первый столбец матрицы справа это вектор i который преобразуем матрицей слева, то же самое и для второго столбца. Определитель, это площадь квадрата между векторами базиса или же цифра, что показывает насколько изменилась площадь единичного базиса после преобразования. На картинке хорошо видно не только это, но и то, почему детерминатор можно вычислить умножением диагонали если матрица треугольная:  
![Картинка](Determinant.PNG)

Отсюда и выходит, что когда определитель равен нулю, то это значит, наши векторы линейно зависимы и значит, что мы теряем ось. Что бы понять сколько осей было потеряно можно вычислить Ранг матрицы. Он показывает сколько осей осталось после преобразования:  
![Картинка](Determinant2.PNG)

Если определитель отрицательный, это означает, что мы как бы условно повернули наше пространство (условно как перевернуть лист бумаги другой стороной). Ещё это можно назвать "инверсией ориентации". Логично что площадь останется такой же:  
![Картинка](Determinant3.PNG)

Зная всё вышесказанное, линейные уравнения можно смело представлять как нахождение начальных координат вектора после преобразования:  
![Картинка](Linear_equation.PNG)

Обратная матрица - это такая матрица, что вернёт вектор после преобразования обратно на свои прошлые координаты. Отсюда и выходит правило, что перемножение матрицы и обратной матрицы даёт единичную матрицу:  
![Картинка](Ones_matrix.PNG)

Нулевой вектор или ядро матрицы - если наше пространство теряет ось, то некоторые векторы сжимаются в начало координат, в точку. И наоборт соотвественно. Отсюда выходит, что если некоторый преобразованный вектор становится нулевым, значит до преобразования он был одним из множества векторов на потерянной оси (не базиса!):  
![Картинка](Zero_space.PNG)

Если у нас допустим матрица $[1 2]$, то это означает, что мы берём 2 измерения и находим куда попадёт вектор на линии с 1 измерением, как на картинке. То есть неквадратные матрицы всё так же показывают преобразования, но как будто ограничивают сами себя или наоборот выходят за рамки существующих осей. Или если не заморачиваться, неквадратные матрицы просто преобразуют входящий вектор в большую или меньшую размерность:  
![Картинка](Non_square_matrices.PNG)
