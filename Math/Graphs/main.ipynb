# Графы

**Граф** — это набор вершин и рёбер, которые их соединяют

Обозначение: $G=(V,E)$, где $V$ — множество вершин, $E$ — множество рёбер  

Пример в нейросети: нейроны — вершины, связи — рёбра  

## Виды графов

**Неориентированный граф**  
Рёбра не имеют направления. Связь «туда-обратно» одинаковая  
![Картинка](An_undirected_graph.jpeg)  
Пример: дружба в соцсети (если A друг B, то B друг A)  

**Ориентированный граф (орграф)**  
Рёбра имеют направление (стрелки)  
![Картинка](A_directed_graph.jpg)  
Пример: подписки в Instagram (A подписан на B, но B может не быть подписан на A)  

**Взвешенный граф**  
У каждого ребра есть число (вес)  
![Картинка](Weighted_graph.png)  
Пример: карта дорог (вес = расстояние между городами)  

**Двудольный граф**  
Вершины делятся на две группы, и рёбра идут только между группами, не внутри  
![Картинка](A_bipartite_graph.webp)  
Пример: пользователи и фильмы (пользователь оценил фильм)  

## Основные понятия

| Термин | Что это | Пример |
|:---|:---|:---:|
| Степень вершины | Сколько рёбер выходит из вершины | Из вершины $A$ выходит 3 линии, значит степень $A = 3$ |
| Путь | Последовательность вершин, соединённых рёбрами | $A \to B \to C$ |
| Цикл | Путь, который начинается и заканчивается в одной вершине | $A \to B \to C \to A$ |
| Связный граф | Из любой вершины можно дойти до любой другой | Иными словами граф один и нет одиноких вершин (ориентированный граф можно считать связным если такой же неориентированный граф связный) |

Заметка: ориентированные графы бывают разной степени связоности (слабо связный, односторонне связный, сильно связный)  

## Дерево — особый граф

Дерево — это связный граф без циклов

Свойства:

- Между любыми двумя вершинами ровно один путь
- Если $n$ вершин, то рёбер ровно $n−1$
- Добавишь одно ребро → появится цикл. Уберёшь одно ребро → потеряешь связность

## DAG — ориентированный ациклический граф
DAG (Directed Acyclic Graph) — ориентированный граф без циклов

Свойства:

- Есть вершины, в которые ничего не входит (источники)
- Есть вершины, из которых ничего не выходит (стоки)
- Можно упорядочить вершины так, что все рёбра идут «слева направо» или «справа налево»

## Способы представления

### Матрица смежности

Берём 2 вершины и проверяем есть ли между ними ребро, если есть пишем 1, иначе 0

Пример:  
|  | A | B | C |
|:---|:---|:---:|:---:|
| A | 0 | 1 | 1 |
| B | 1 | 0 | 0 |
| C | 1 | 0 | 0 |

Нюансы:
- Если граф ориентированный, то значение слева должно идти в значение сверху (ну или значение строки, в значение столбца). То есть пропадает зеркальность, в отличии от прошлого раза
- Если в вершине $A$ петля, то из $A$ в $A = 1$
- Если рёбер несколько, то ставится не 1, а количество рёбер
- Если взвешенный граф, то ставим вес рёбер

### Список смежности

Для каждой вершины — список её соседей

Пример:  
A: [B, C]  
B: [A]  
C: [A]  

Нюансы такие же как у матрицы. Если есть веса, то пишем в скобках. Остальное в целом легко показать не используя ничего дополнительно, просто грамотно расписываем списки у вершин

## Обходы графа

Это способы найти все вершины. Важно что в любом из двух способов мы должны помечать точки которые уже посетили

**Обход в глубину (DFS)** - осуществляется через очередь. Первый элемент в очереди отмечается как пройденый, а непосещённые его соседи закидываются в конец очереди и так по циклу (Как круги на воде)

**Обход в ширину (BFS)** - осуществляется через стек или рекурсию. Мы идём максимально глубоко пока не упрёмся в тупик. Как только упёрлись, записали вершину и возвращаемся назад проверяя следующие возможные пути и так по циклу (Как в лабиринте)

## Кратчайшие пути

### Алгоритм Дейкстры

**Алгоритм Дейкстры** — это способ найти кратчайшие пути от одной вершины до всех остальных во взвешенном графе, где веса рёбер неотрицательные(иначе алгоритм сломается)  

Это тот алгоритм, где мы для всех непосещённых вершин ставим бесконечность, а после посещения дописываем кратчайшие пути. Наглядный пример есть в папке "dijkstra_steps"
