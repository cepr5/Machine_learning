# Основные статистические понятия

*Пример, оценки 10 студентов за экзамен: {3,4,4,5,5,5,6,7,9,10}. Этот пример будет использоваться ниже*

## Среднее значение

**Среднее значение (Mean)** - это сумма всех значений, делённая на их количество
$$
\bar{x} = \frac{1}{n}\sum_{i=1}^{n}x_{i}
$$
где n - всего объектов в множестве, x - значение конкретного объекта

В нашем примере:
$$
\bar{x} = \frac{3+4+4+5+5+5+6+7+9+10}{10} = 5.8
$$

## Медиана

**Медиана (Median)** - это значение посередине **отсортированного** ряда

- если n нечётное, берём центральный элемент
- если n четное, берём среднее двух центральных

В нашем примере n четное и две цифры посередине это 5 и 5. Их средняя 5 - это и есть медиана

## Мода

**Мода (Mode)** - это значение, которое встречается чаще всего

В нашем примере 5 встречается 3 раза, что самое частое. Значит мода 5

*Мода не обязательно одно число, это может быть и вектор чисел!*

## Дисперсия

**Дисперсия (Variance)** - это мера разброса. Показывает, насколько значения «размазаны» вокруг среднего
$$
Var(X) = \frac{1}{n} \sum_{i=1}^{n} (x_{i} - \bar{x})^{2}
$$

В нашем примере:
$$
Var(X) = \frac{(3 - 5.8)^2 + (4 - 5.8)^2 + (4 - 5.8)^2 + (5 - 5.8)^2 + (5 - 5.8)^2 + (5 - 5.8)^2 + (6 - 5.8)^2 + (7 - 5.8)^2 + (9 - 5.8)^2 + (10 - 5.8)^2}{10} = 4.56
$$

## Стандартное отклонение

**Стандартное отклонение (Std)** - это квадратный корень из дисперсии. Возвращает разброс в тех же единицах, что и данные. Или простым языком, в отличии от дисперсии он показывает буквальный диапозон разброса между средней
$$
\sigma = \sqrt{Var(X)}
$$

В нашем случаи:
$$
\sigma = \sqrt{4.56} \approx 2.14
$$
Стандартное отклонение $\sigma \approx 2.14$ означает, что в среднем оценки отклоняются от среднего (5.8) примерно на 2 балла. Если данные нормальны, то ~68% значений лежат в диапазоне $5.8 \pm 2.14 = [3.66; 7.94]$

## Вероятность

**Вероятность (Probability)** - это число от 0 до 1, показывающее шанс события

Пример: если честно бросить кубик 1000 раз, доля шестёрок будет $\approx \frac{1}{6}$. То есть классическая вероятность из дискретки

## Распределения

**Распределения (Distributions)** - это функция, показывающая, какие значения принимает переменная и с какой вероятностью

Примеры: Бернулли, бином, Пуассон в дискретной вероятности и есть другие в непрерывной вероятности (изучу далее)

## Корреляция

**Корреляция (Correlation)** - это мера линейной связи между двумя переменными. Число от -1 до 1
$$
P_{XY} = \frac{Cov(X,Y)}{\sigma_X*\sigma_Y}
$$

- если $p = 1$ $\to$ идеальная линейная связь, Y растёт вместе с X
- если $0 < p < 1$ $\to$ линейная связь есть, но не идеальная (разброс)
- если $p = 0$ $\to$ линейной связи нет (но может быть нелинейная!)
- если $-1 < p < 0$ $\to$ линейная связь есть, но не идеальная, и Y убывает при росте X
- если $p = -1$ $\to$ идеальная линейная связь, Y убывает при росте X

## Регрессия

**Регрессия (Regression)** - это метод моделирования зависимости одной переменной от других

Простая линейная:
$$
y = w_0 + w_1x
$$

Множественная:
$$
y = w_0 + w_1x_1 + w_2x_2 + ... + w_nx_n
$$
