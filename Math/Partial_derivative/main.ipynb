# Частная производная

Если обычная производная получается вот так:  
Функция:  
$$ 
f(x) = x^2 
$$  
Производная:  
$$ 
\frac{df}{dx} = 2x 
$$  
---
Частная производная, это когда у нас две переменных и мы берём производную по каждой переменной отдельно, пока все остальные становятся константами (не буквально, а ведут себя как константы в контексте нахождения производной):  
Функция:  
$$
f(x,y) = x^2 + y^2
$$
Производная по $x$, а $y$ как константа:  
$$
\frac{\partial f}{\partial x} = 2x
$$
Производная по $y$, а $x$ как константа:  
$$
\frac{\partial f}{\partial y} = 2y
$$
---
Более сложный пример:  
Функция:  
$$
f(x,y) = x^2y + 3xy^2 + 5
$$
Производная по $x$, а $y$ как константа:  
$$
\frac{\partial f}{\partial x} = 2xy + 3y^2
$$
Производная по $y$, а $x$ как константа:  
$$
\frac{\partial f}{\partial y} = x^2 + 6xy
$$
---

# Градиент

Это вектор из частных производных, который показывает направление роста функции. Если простым языком, то условно это можно представить как допустим передвижение в двухмерном пространстве. Мы можем идти вперёд/назад или влево/вправо. Мы допустим хотим спустится с какого-то склона, который как раз посчитан функцией. Сама функция это функция высот дпоустим. Для этого мы находим частные производные по каждой оси и смотри куда функция будет уменьшаться. Допустим получился вектор [вперёд влево], значит так и двигаемся, что бы спускаться:
$$
\nabla L = \left( \frac{\partial L}{\partial w_1}, \frac{\partial L}{\partial w_2}, \dots, \frac{\partial L}{\partial w_n} \right)
$$
