## Транспонирование

Строки становится столбцами, а столбцы строками  
$A = [[1, 2, 3], [4, 5, 6]]$  
$Aᵀ = [[1, 4], [2, 5], [3, 6]]$

`A = A.T`

## Обратная матрица

$A · A⁻¹ = A⁻¹ · A = I$, где I - единичная матрица

Условия существования:  
- Квадратная матрица
- Определитель != 0

`A = np.array([[1, 2], [3, 4]])`  
`A_inv = np.linalg.inv(A)`

## Псевдообратная матрица

*Когда обратной матрицы не существует из-за того, что она не квадратная или вырожденная(Определитель == 0)*  

A размер (m, n)  
A⁺ размер (n, m)

Вычисление через SVD:  
$A = U · Σ · Vᵀ$  
$A⁺ = V · Σ⁺ · Uᵀ$  
где Σ⁺ получается заменой всех ненулевых сингулярных чисел σ_i на 1/σ_i, а нулевые остаются нулями.  

*Если A квадратная и невырожденная, то A⁺ = A⁻¹*  

`C = np.array([[1, 2],[3, 4]])`  
`C_pinv = np.linalg.pinv(C)`

## Определитель

Число, показывающее, во сколько раз матрица масштабирует объём, и её обратимость  
Только для квадратных матриц  

`A = np.array([[1, 2], [3, 4]])`  
`det_A = np.linalg.det(A)`  
`if det_A == 0: "Матрица вырожденная, обратной матрицы нету"`

*Определитель может быть не нулём, но очень к нему близок, поэтому рекомендуется сравнивать с очень малым числом*
