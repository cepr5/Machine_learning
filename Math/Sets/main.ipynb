# Множества

Множество — это совокупность объектов, рассматриваемая как единое целое  
Примеры:  
$$
\{1,2,3\} \text{ - три числа}
$$
$$
\{a, b, c\} \text{ - три буквы}
$$
$$
\{1, 2, 3, 4, ...\} \text{ - множество Натуральных чисел (N)}
$$
$$
\text{и т.д.}
$$

Ключевые моменты:

- порядок не важен $\{1,2,3\} = \{2,3,1\}$
- повторы не считаются $\{1,1,2,3\} = \{1,2,3\}$
- пустое множество (в котором 0 элементов) обозначается: $\varnothing$

## Принадлежность элемента

Если элемент принадлежит множеству, тогда: $2 \in \{1,2,3\}$  
Если не принадлежит, тогда: $x \notin \{1,2,3\}$  

## Стандартные числовые множества

**Натуральные числа** ($\mathbb{N}$) — $\{1, 2, 3, 4, \dots\}$  
**Целые числа** ($\mathbb{Z}$) — $\{\dots, -2, -1, 0, 1, 2, \dots\}$  
**Рациональные числа** ($\mathbb{Q}$) — $\left\{\frac{p}{q} \mid p, q \in \mathbb{Z}, q \neq 0\right\}$  
**Иррациональные числа** ($\mathbb{I}$) — $\mathbb{R} \setminus \mathbb{Q}$ (нельзя записать дробью, десятичная запись бесконечна и непериодична. Например: $\pi, e, \sqrt{3} = 1.7320508075688772...$)  
**Вещественные числа** ($\mathbb{R}$) — $\mathbb{Q} \cup \mathbb{I}$  
**Комплексные числа** ($\mathbb{C}$) — $\{a + bi \mid a, b \in \mathbb{R}, i^2 = -1\}$  

## Типы множеств

Конечные - $\{1,2,3\}$  
Счётные - $\mathbb{N}, \mathbb{Z}, \mathbb{Q}$  
Несчётные - $\mathbb{I}, \mathbb{R}, \mathbb{C}$  
(В классическом определении: если элементы множества можно перенумеровать натуральными числами (1, 2, 3, …) — оно счётное. Если нельзя — несчётное. Тут этот момент просто ломает мозг. Думаю эта тема станет понятнее если вникнуть в тему бессконечности чуть глубже, а пока можно просто запомнить по **стандартным числовым множествам**)  

## Операции

$A = \{1,2,3\}$  
$B = \{3,4,5\}$

**Объединение** - из двух множеств выписываются все значения, исключая повторы:  
$A \cup B = \{x|x \in A \text{ или } x \in B\}$  
$A \cup B = \{1,2,3,4,5\}$  
**Пересечение** - из двух множество выписываются только те элементы, что встречаются и в первом и во втором множестве:  
$A \cap B = \{x|x \in A \text{ и } x \in B\}$  
$A \cap B = \{3\}$  
**Разность** - всё что есть в $А$, но нет в $В$:  
$A \setminus B = \{x|x \in A \text{ и } x \notin B\}$  
$A \setminus B = \{1, 2\}$  
**Симметрическая разность** - всё, что есть только в одном из множеств:  
$A \triangle B = (A \setminus B) \cup (B \setminus A)$  
$A \triangle B = \{1,2,4,5\}$  
**Дополнение** - всё, чего нет в $A$ (относительно универсального $U$):  
$U = \{1,2,3,4,5\}$  
$A^c = U \setminus A$  
$A^c = \{4,5\}$  
**Декартово произведение** - все упорядоченные пары $(a, b)$:  
$A = \{1,2\}$  
$B = \{x,y\}$  
$A \times B = \{(a,b) | a \in A, b \in B\}$  
$A \times B = \{(1,x),(1,y),(2,x),(2,y)\}$  

## Подмножества  

$A$ — подмножество $B$, если каждый элемент $A$ есть в $B$:  
$A = \{1,2\}$  
$B = \{1,2,3\}$  
$A \subseteq B$  
$B \not\subset A$  
(Чёрточка внизу значка подразумевает, что множество принадлежит или равно. Если мы точно знаем, что только принадлежит, но не равно, то пишем без нижней чёрточки)

## Мощность

Это просто количество элементов в множестве:  

$A = \{a,b,c\}$  
$|A| = 3$  
Если $A = \varnothing$, то:  
$|A| = 0$

## Законы

Эти законы выполняются для любых множеств:

**Коммутативность**  
$A \cap B = B \cap A$  
$A \cup B = B \cup A$  

**Ассоциативность**  
$(A \cap B) \cap C = A \cap (B \cap C)$  
$(A \cup B) \cup C = A \cup (B \cup C)$  

**Дистрибутивность**  
$A \cap (B \cup C) = (A \cap B) \cup (A \cap C)$  
$A \cup (B \cap C) = (A \cup B) \cap (A \cup C)$  

**Идемпотентность**  
$A \cap A = A$  
$A \cup A = A$  

**Поглощение**  
$A \cap (A \cup B) = A$  
$A \cup (A \cap B) = A$  

**Де Морган**  
$(A \cap B)^c = A^c \cup B^c$  
$(A \cup B)^c = A^c \cap B^c$
