# Дискретная вероятность

Дискретная вероятность — это теория вероятностей на конечных или счётных множествах исходов. Всё считается через суммы, а не интегралы

## Пространство исходов и событие

**Пространство исходов** — это «список всего, что может случиться»  

Пространство исходов обозначается буквой $\Omega$ (омега)  

Пример: при броске шестигранного кубика пространство исходов будет: $\Omega = \{1, 2, 3, 4, 5, 6\}$  
Или например, бросание двух монет: $\Omega = \{OO, OP, PO, PP\}$  
Не обязательно оно обязано быть простым, к примеру, пространство исхода выпадения первого Орла: $\Omega = \{O, PO, PPO, PPPO, ...\}$  

**Элементарный исход** — это один конкретный результат. Например, на кубике выпало 5  

Если оно принадлежит к пространству исходов $\Omega$, то обозначается как $\omega$  

**Событие** — это «подмножество Пространства исходов (или набор элементарных исходов), которое нас интересует»  

Обозначается обычно заглавными буквами: $A$ или $B$ или $C$ и т.д.  

Пример с кубиком: $\Omega = \{1, 2, 3, 4, 5, 6\}$  
Событие выпадение чётного числа: $A = \{2, 4, 6\}$  
*Событие происходит, если результат эксперимента попал в это подмножество*  

## Вероятность события

Обозначается как P(_). Где в скобках указываем вероятность чего именно мы имеем ввиду  

Вероятность элементарного исхода всегда находится в диапазоне:  

$$
0 \le P(\omega) \le 1
$$

А сумма вероятностей всех исходов, равна 1:

$$
\sum_{\omega \in \Omega} P(\omega) = 1
$$

Вероятность события $A$ - это сумма всех входящих в него вероятностей исходов:  

$$
P(A) = \sum_{\omega \in A} P(\omega)
$$

Если все исходы **равновероятны**, формула упрощается:  

$$
P(A) = \frac{|A|}{|\Omega|}
$$
Где $|A|$ и $|\Omega|$ - это сумма числа исходов (не вероятностей, а просто количество)

## Виды событий

$A \cup B$ *- произошло хотя бы одно из событий*  
$A \cap B$ *- произошли оба события*

**Достоверное событие** - происходит всегда и равно 1 

$$
P(\Omega) = 1
$$  

**Невозможное событие** - событие, которое никогда не произойдёт и равно 0 

$$
P(\varnothing) = 0
$$

**Противоположное событие** - обозначается как $\overline{A}$ или $A^c$. Это все оставшиеся исходы, которые не входят в $A$

$$
P(\overline{A}) = 1 - P(A)
$$

$$
P(\overline{A}) + P(A) = P(\Omega) = 1
$$

**Несовместные события** - те события, которые не могут произойти одновременно

$$
A \cap B = \varnothing
$$

**Объединение для несовместных**. То есть если оба множества не содержат одинаковых элементов, то их можно суммировать:

$$
P(A \cup B) = P(A) + P(B)
$$

**Общая формула включений-исключений**. Для ситуаций, когда множества могут содержать одинаковые исходы:

$$
P(A \cup B) = P(A) + P(B) - P(A \cap B)
$$

Такая формула означает: функция $P$ берёт событие из $F$ и возвращает число от 0 до 1. Где $F$ это набор всех событий из $\Omega$, о которых мы можем говорить.
$$
P : F \to [0,1]
$$

## Аксиомы вероятности (Колмогоров)

Аксиомы Колмогорова — это три правила, которые определяют, что такое вероятность. Всё остальное — следствия  

**Аксиома 1: Неотрицательность** - вероятность любого события не может быть отрицательной
$$
P(A) \ge 0
$$
**Аксиома 2: Нормированность** - вероятность всего пространства исходов равна 1
$$
P(\Omega) = 1
$$
**Аксиома 3: Аддитивность (счётная)** - если события не могут произойти одновременно, то вероятность того, что произойдёт хотя бы одно из них, равна сумме их вероятностей
$$
P(A_1 \cup A_2 \cup A_3 \cup ...) = P(A_1) + P(A_2) + P(A_3) + ...
$$

Теоремы, которые выводятся из аксиом:

1. Вероятность невозможного события равна 0 (было выше)
2. Вероятность противоположного события (было выше)
3. Вероятность не больше 1 (было выше)
4. Монотонность (если $A \subseteq B$, то $P(A) \le P(B)$)
5. Формула включений-исключений (было выше)
6. Непрерывность. Если $A_1 \subseteq A_2 \subseteq A_3 \subseteq ...$, то:
$$
P\left(\bigcup_{n=1}^{\infty} A_n\right) = \lim_{n \to \infty} P(A_n)
$$

## Условная вероятность

**Условная вероятность** — это вероятность события при условии, что уже произошло другое событие (при условии, что $P(B) > 0$):  
$$
P(A|B) = \frac{P(A) \cap P(B)}{P(B)}
$$  
или
$$
P(A|B) = \frac{|A \cap B|}{|B|}
$$  
Читается: «вероятность $A$ при условии $B$» или «вероятность $A$, если известно, что $B$ произошло»

Пример с кубиком:  
Событие $A$: «выпало чётное» $= \{2, 4, 6\}$  
Событие $B$: «выпало больше 3» $= \{4, 5, 6\}$  
$P(A)$ = $P(B) = 3/6$  
$P(A) \cap P(B) = 2/6$ (исходы 4 и 6 из $\Omega$)  
$P(A|B) = \frac{2/6}{3/6} = \frac{2}{3}$   
*Это вероятность того, что произойдёт событие $A$ после того, как произошло событие $B$*

Если два события независимы, то: $P(A|B) = P(A)$. То есть событие $B$ никак не влияет на событие $A$ если они не пересекаются

**Формула умножения** - вероятность, что произойдут и $A$, и $B$, равна вероятности $B$, умноженной на вероятность $A$ при условии $B$:
$$
P(A \cap B) = P(B) * P(A|B)
$$
или
$$
P(A \cap B) = P(A) * P(B|A)
$$

## Формула полной вероятности

Если события $B_1, B_2, ... B_n$ образуют разбиение $\Omega$ (то есть не пересекаются и в объединении дают всё $\Omega$), то:
$$
P(A) = \sum_{i=1}^{n} P(A|B_i) * A(B_i)
$$

Пример: на двух заводах делают лампочки
- Завод 1: 60% продукции, 2% брака
- Завод 2: 40% продукции, 5% брака

Случайно берём лампочку. Какова вероятность, что она бракованная P(A)?

Разбор:  
Общее количество лампочек это $\Omega$  
Вероятность того, что Завод 1 соберёт рандомную 1 лампочку 60%  
Вероятность того, что Завод 2 соберёт рандомную 1 лампочку 40%  
Они не пересекаются и в общем дают всю сумму $\Omega$  
Есть некоторое событие А, которое говорит, что лампочка бракованна  
В случаи если лампочка была сделана на Заводе 1 (событие которое уже произошло), то шанс события А 2%  
В случаи если лампочка была сделана на Заводе 2 (событие которое уже произошло), то шанс события А 5%  
Отсюда и выходит что:
$$
P(A) = P(A | Завод 1) * P(Завод 1) + P(A | Завод 2) * P(Завод 2)
$$
$$
P(A) = 0.02 * 0.6 + 0.05 * 0.4 = 0.032
$$

## Формула Байеса

Формула Байеса позволяет перевернуть условную вероятность. Зная P(A|B), можно найти P(B|A). Где знаменатель не равен нулю:
$$
P(B|A) = \frac{P(A|B) * P(B)}{P(A)}
$$

Пример. Какова вероятность, что после положительного теста человек действительно болен?  
B = "Болен"  
Z = "Здоров"  
"Тест положительный" = +  

$P(B) = 0.01$  
$P(Z) = 0.99$  
$P(+|B) = 0.95$  
$P(+|Z) = 0.05$ (ложное срабатывание)  

Мы не знаем P(+), считаем через полную вероятность:  
$P(+) = 0.01 * 0.95 + 0.99 * 0.05 = 0.059$

Применяем формулу Бейеса:  
$P(B|+) = \frac{0.95 * 0.01}{0.059} = 0.161$  

*Всё верно, т.к. болезнь в целом редкая (всего 1%) то это значит, что тест слишком часто даёт ложные срабатывания и по итогу действительно больных по положительному тесту всего 16%*

Вот как выглядит дерево решения:  
```text
              ┌── + (0.95)  → 0.01 × 0.95 = 0.0095  ← настоящий плюс
   ┌─ болен (0.01)
   │          └── − (0.05)  → 0.01 × 0.05 = 0.0005
───┤
   │          ┌── + (0.05)  → 0.99 × 0.05 = 0.0495  ← ложный плюс
   └─ здоров (0.99)
              └── − (0.95)  → 0.99 × 0.95 = 0.9405
```

Полная формула. Для случая когда есть множество $\{B_1, B_2, ... B_n\}$ где все элементы не пересекаются и в сумме дают $\Omega$ или $1$. В знаменателе просто *Формула полной вероятности* для $A$:
$$
P(B_k|A) = \frac{P(A|B_k) * P(B_k)}{\sum_{i=1}^n P(A|B_i) * P(B_i)}
$$

Формула в виде «шансов»:
$$
\frac{P(B|A)}{P(\overline{B}|A)} = \frac{P(A|B)}{P(A|\overline{B})} * \frac{P(B)}{P(\overline{B})}
$$
Смысл: апостериорные шансы = отношение правдоподобий × априорные шансы (Допустим)

## Независимость

Если события не влияют друг на друга, то они независимы
$$
P(A \cap B) = P(A) * P(B)
$$
$$
P(A|B) = P(A)
$$

Если события точно никак независимы можно применить следующую формулу. Но тут главное ПОЛНАЯ независимость событий. К примеру бросок 3ёх разных монет. Они точно никак друг на друга не влияют и поэтому их события можно умножать друг с другом:  
$$
P(A_1 \cap A_2 \cap A_3 \cap ...) = P(A_1) * P(A_2) * P(A_3) * ...
$$

*Важно понимать разницу между независимостью и несовместностью. Первая говорит о влиянии событий друг на друга, а вторая, пересекаются ли их элементарные исходы*

## Закон распределения

|x_i|x_1|x_2|...|x_n|
|-|-|-|-|-|
|p_i|p_1|p_2|...|p_n|

Где x - элементарный исход, p - вероятность исхода, а так же:
$$
\sum_{x \in X} P(x) = 1
$$

## В теории вероятности не множества, а функция

То есть очень часто будет запись, примерно, типа: $P(X = k)$. Где $X$ функция которая ставит каждому $\omega$ его соотвествующее значение. Например:  
Бросаем две монеты, все $\Omega$ = {OO, PO, OP, PP}  
Если X = числу орлов, то:
| Результат | Число Орлов |
| - | - |
| OO | 2 |
| PO | 1 |
| OP | 1 |
| PP | 0 |

В таком случаи $P(X = 1) = \frac{2}{4}$. Потому что из всех возможных событий, только 2 соотвествует тому, что выпал 1 Орел

## Функция распределения (CDF)

**Функция распределения** показывает сколько всего вероятности накопилось до $x$:
$$
F(x) = P(X \le x)
$$
или
$$
F(x) = \sum_{x_i \le x} p_i
$$

Пример графика для кубика, где 6 исходов и у всех шансы $\frac{1}{6}$:

```text
F(x)
 1  ┤                                ┌────────
    │                                │
5/6 ┤                          ┌─────┘
    │                          │
4/6 ┤                    ┌─────┘
    │                    │
3/6 ┤              ┌─────┘
    │              │
2/6 ┤        ┌─────┘
    │        │
1/6 ┤  ┌─────┘
    │  │
 0  ┼──┴─────┴─────┴─────┴─────┴─────┴──────→ x
      1     2     3     4     5     6
```

## Математическое ожидание

Математическое ожидание — это среднее значение случайной величины, если повторять эксперимент много-много раз. События должны быть несовместными
$$
E[X] = x_1p_1 + x_2p_2 + ... + x_np_n
$$
$$
E[X] = \sum_{i=1}^n x_ip_i
$$
Например, крутим рулетку 5 раз. Начальная ставка всегда 100. А при победе 5 раз подряд получим 3200 (победа даёт х2), но шанс всего $\frac{1}{32}$ (чёрное и белое с шансом 50%, все комбинации дадут $2^5$). Таким образом:
$$
EV = 3200 * \frac{1}{32} + 0 * \frac{31}{32}  = 100
$$
То есть по итогу выходим в ноль при честной игре в 50%. В этом варианте не учитывается, что мы что-то отдаём. Как будто бесплатно играем, поэтому можно дополнительно вычесть 100  
Есть второй способ заполнения формулы:  
$$
EV = (3200 - 100) * \frac{1}{32} + (0-100) * \frac{31}{32}  = 0
$$

## Дисперсия

**Дисперсия** — это мера разброса случайной величины вокруг её среднего (матожидания). Или другим языком - это насколько сильно значения отклоняются от среднего
$$
D[X] = \sum_{i=1}^n (x_i - E[X])^2 * p_i
$$

Пример для кубика:

$X$ — выпавшее число на честном кубике.

Значения: $1, 2, 3, 4, 5, 6$.  
Вероятности: $p_i = \frac{1}{6}$ для каждого.

**Шаг 1. Матожидание**

$$
E[X] = 1 \cdot \frac{1}{6} + 2 \cdot \frac{1}{6} + 3 \cdot \frac{1}{6} + 4 \cdot \frac{1}{6} + 5 \cdot \frac{1}{6} + 6 \cdot \frac{1}{6}
$$

$$
E[X] = \frac{1+2+3+4+5+6}{6} = \frac{21}{6} = 3.5
$$

**Шаг 2. Дисперсия по определению**

$$
D[X] = \sum_{i=1}^{6} (i - 3.5)^2 \cdot \frac{1}{6}
$$

Таблица отклонений:

| $x_i$ | $x_i - 3.5$ | $(x_i - 3.5)^2$ | $\cdot \frac{1}{6}$ |
|---|---|---|---|
| 1 | −2.5 | 6.25 | 1.0417 |
| 2 | −1.5 | 2.25 | 0.375 |
| 3 | −0.5 | 0.25 | 0.0417 |
| 4 | 0.5 | 0.25 | 0.0417 |
| 5 | 1.5 | 2.25 | 0.375 |
| 6 | 2.5 | 6.25 | 1.0417 |

Сумма:

$$
D[X] = 1.0417 + 0.375 + 0.0417 + 0.0417 + 0.375 + 1.0417 \approx 2.917
$$

**Альтернативная формула дисперсии**

$$
D[X] = E[X^2] - (E[X])^2
$$

Считаем $E[X^2]$:

$$
E[X^2] = \frac{1^2 + 2^2 + 3^2 + 4^2 + 5^2 + 6^2}{6} = \frac{1+4+9+16+25+36}{6} = \frac{91}{6} \approx 15.1667
$$

Квадрат матожидания:

$$
(E[X])^2 = 3.5^2 = 12.25
$$

Дисперсия:

$$
D[X] = 15.1667 - 12.25 = 2.9167
$$

**Шаг 3. Стандартное отклонение**

$$
\sigma = \sqrt{D[X]} = \sqrt{2.9167} \approx 1.708
$$

Стандартное отклонение означает, что большинство значений лежит в диапазоне $3.5 \pm 1.71$, то есть примерно от $1.79$ до $5.21$.

## Основные дискретные распределения

**Равномерное дискретное** - когда все значения равновероятны. То есть самые банальные как монетка и кубик
$$
P(X = x_i) = \frac{1}{n}
$$
Матожидание:
$$
E[X] = \frac{1}{n}\sum x_i
$$

**Распределение Бернулли** - один эксперимент с двумя исходами: успех или неудача
$$
P(X = 1) = p
$$
$$
P(X = 0) = 1 - p
$$
Матожидание:
$$
E[X] = p
$$
Дисперсия:
$$
D[X] = p(1 - p)
$$

**Биномиальное распределение** - $n$ независимых испытаний Бернулли с одинаковой вероятностью успеха $p$. Считаем число успехов  
В следующей формуле получается, что:
- n - общее число испытаний (сколько раз проводим эксперимент)
- k - число успехов (сколько раз из $n$ произошёл успех)
- p - вероятность успеха в одном испытании
- P(X = k) - вероятность, что успехов будет ровно $k$. Или простым языком: "Сколько вероятность, что из n испытаний, будет успех k раз"
- в скобках сочетание из комбинаторики
$$
P(X = k) = \binom{n}{k}p^k(1 - p)^{n - k}
$$
Матожидание:
$$
E[X] = np
$$
Дисперсия:
$$
D[X] = np(1 - p)
$$

**Геометрическое распределение** - повторяем испытания Бернулли до первого успеха. Считаем число попыток (включая успешную) или число неудач до успеха

$$
P(X = k) = (1 - p)^{k-1} * p
$$
где k - номер испытания на котором успех после поражений

Матожидание:
$$
E[X] = \frac{1}{p}
$$
Дисперсия:
$$
D[X] = \frac{1 - p}{p^2}
$$

Пример, успех на выпадение 6 на кубике:
- $p = \frac{1}{6}$
- $q = 1 - \frac{1}{6} = \frac{5}{6}$

Хотим событие: первая шестёрка выпала на 4-м броске.
Это значит: 3 раза выпало не 6, а на 4-й раз выпало 6

$$
P(X = 4) = \frac{5}{6}^{4-1} * \frac{1}{6} \approx 0,096 \approx 9\%
$$

*Важно! Когда допустим мы задаём k=4 на выпадение Орла, то это строго лишь одна комбинация: PPPO*  
*Ещё момент. Важно не путать с условной вероятностью. То есть к примеру: P(O|PPP) говорит о вероятности Орла когда х3 Решки уже произошли. В нашем же случаи мы считаем всё вместе. Узнаем каков вообще шанс, что выпадут х3 Решки и ещё Орёл после них*

**Распределение Пуассона** - считаем число редких событий за фиксированный интервал времени, длины, площади и т.д. (или простым языком, мы берём среднее значение в каком-то промежутке и в этом же промежутке проверяем вероятность конкретного числа)

Другое определение: задан интервал (час, день, страница) и средняя интенсивность $λ$ — сколько событий в среднем происходит за этот интервал. Формула $P(X=k)$ даёт вероятность, что за тот же интервал произойдёт ровно $k$ событий

$$
P(X = k) = \frac{\lambda^ke^{-\lambda}}{k!}
$$

Матожидание:
$$
E[X] = \lambda
$$
Дисперсия:
$$
D[X] = \lambda
$$

Пример, в колл-центре в среднем за день 3 звонка в час ($\lambda = 3$)

Вероятность 0 звонков за день:
$$
P(X = k) = \frac{3^0e^{-3}}{0!} \approx 0.0498 \approx 5\%
$$
Вероятность 3 звонков за день:
$$
P(X = k) = \frac{3^3e^{-3}}{3!} \approx 0.224 \approx 22\%
$$
Вероятность 10 звонков за день:
$$
P(X = k) = \frac{3^10e^{-3}}{10!} \approx 0.00081 \approx 0\%
$$

**Гипергеометрическое распределение** - выборка без возвращения из конечной популяции. Считаем число «успехов» в выборке
$$
P(X = k) = \frac{\binom{K}{k} * \binom{N - K}{n - k}}{\binom{N}{n}}
$$

Матожидание:
$$
E[X] = n * \frac{K}{N}
$$
Дисперсия:
$$
D[X] = n * \frac{K}{N} * (1 - \frac{K}{N}) * \frac{N - n}{N - 1}
$$

Пример: Есть колода из 36 карт. В ней 4 туза. Мы вытаскиваем 5 карт наугад (без возвращения). Какова вероятность, что среди них окажется ровно 2 туза?
- N - всего карт в колоде (36)
- K - «успешных» карт (тузов) (4)
- n - сколько вытаскиваем (5)
- k - сколько тузов хотим (2)

$$
P(X = k) = \frac{\binom{4}{2} * \binom{36 - 4 = 32}{5 - 2 = 3}}{\binom{36}{5}} \approx 0.0789 \approx 8\%
$$

## Совместное распределение

**Совместное распределение** — это когда мы смотрим на две или больше случайных величин одновременно. Часто знание величины даёт информацию о другой. Совместное распределение описывает всю картину целиком: как они ведут себя вместе

Самый простой пример, это когда берём два кубика (независимые величины). Тогда вероятность допустим выпадения на одном кубике 1, а на втором 2, при одновременном броске обоих будет:
$$
P(X = 1, Y = 2) = \frac{1}{36}
$$

Более интересный пример:  
Бросаем две монеты
- X - сколько всего Орлов (0,1,2)
- Y - Орёл ли на первой монете (0, 1)

*Тут уже две величины зависимы*

Исходы: {OO, PO, PO, PP}

Таблица совместного распределения отлично показывает суть:
| X\Y | Y=0 | Y=1 |
|  -  |  -  |  -  |
| X=0 | 1/4 |  0  |
| X=1 | 1/4 | 1/4 |
| X=2 |  0  | 1/4 |

*Причём вся сумма исходов по итогу должна давать 1. По таблице это опять же наглядно видно*

**Маргинальные распределения** - это способ свернуть совместное распределение до одной величины, суммируя по другой

Для $X$: суммируем по строкам
$$
P(X = x) = \sum_y P(X = x, Y = y)
$$
| X | P(X) |
|-|-|
| 0 | 1/4 |
| 1 | 1/4 + 1/4 = 2/4 |
| 2 | 1/4 |

Для $X$: суммируем по столбцам
$$
P(Y = y) = \sum_x P(X = x, Y = y)
$$
| Y | P(Y) |
|-|-|
| 0 | 1/4 + 1/4 = 2/4 |
| 1 | 1/4 + 1/4 = 2/4 |

**Совместная таблица с маргинальными**
| X\Y  |  0  |  1  | P(X) |
|  -   |  -  |  -  |  -   |
|  0   | 1/4 |  0  | 1/4  |
|  1   | 1/4 | 1/4 | 2/4  |
|  2   |  0  | 1/4 | 1/4  |
| P(Y) | 2/4 | 2/4 |  1   |

### Независимость через совместное распределение

Две величины независимы, если:
$$
P(X = x, Y = y) = P(X = x) * P(Y = y)
$$
Для всех пар $(x,y)$

Проверим наш пример с монетами (пример зависимых):
$$
P(X = 2, Y = 0) = 0
$$
$$
P(X = 2) * P(Y = 0) = \frac{1}{8}
$$

А для двух кубиков (пример независимых):
$$
P(X = 3, Y = 4) = \frac{1}{36}
$$
$$
P(X = 3) * P(Y = 4) = \frac{1}{36}
$$

**Условное распределение из совместного** - то же самое, только формула чуть другая:
$$
P(Y = y | X = x) = \frac{P(X = x | Y = y)}{P(X = x)}
$$

Пример: какова вероятность Y = 1, если известно, что X = 2:
$$
P(Y = 1 | X = 2) = \frac{P(X = 2 | Y = 1)}{P(X = 2)}
$$

**Матожидание функции двух величин** - здесь под $g(X,Y)$ понимается ЛЮБАЯ функция. Примеры: $X + Y$, $X * Y$, $max(X, Y)$ и т.д.
$$
E[g(X,Y)] = \sum_x \sum_y g(x,y) * P(X = x, Y = y)
$$

**Линейность матожидания** - более простой подсчёт для конкретного случая (тут мы можем добавлять множители, квадраты, корни с обоих сторон, например: E(X^2 + Y^2). Но для таких случаев лучше уточнить):
$$
E[X + Y] = E[X] + E[Y]
$$

### Ковариация

**Ковариация** — это мера линейной связи между $X$ и $Y$:
$$
Cov(X,Y) = E[(X - E[X])(Y - E[Y])]
$$

Или через удобную формулу:
$$
Cov(X,Y) = E[XY] - E[X] * E[Y]
$$

Смысл:
- Cov > 0 - когда X растёт, Y тоже растёт
- Cov < 0 - когда X растёт, Y падает
- Cov = 0 - линейной связи нет

Если X и Y независимы, то: 
$$
Cov(X,Y) = 0
$$
*$Cov(X,Y) = 0$ не значит, что независимы. Может быть нелинейная зависимость*

### Корреляция

**Корреляция** - это надстройка над **ковариацией**, которая помогает ответы сложить в промежуток $-1 \le p(X,Y) \le 1$
$$
p(X,Y) = \frac{Cov(X,Y)}{\sigma_X * \sigma_Y}
$$
Где:
- $\sigma_X = \sqrt{D[X]}$
- $\sigma_Y = \sqrt{D[Y]}$

### Дисперсия суммы

Для двух величин, если зависимы:
$$
D[X + Y] = D[X] + D[Y] + 2Cov(X,Y)
$$

Для двух величин, если независимы:
$$
D[X + Y] = D[X] + D[Y]
$$

**Совместное распределение для $n$ величин** - суть тут в том, что может быть не только две величины X и Y, а куда больше. Расcматривать подробно не будем, определим лишь то, что для любой размерности всегда должно выполняться правило, что все вероятности дают 1:
$$
\sum_{x_1} \sum_{x_2} ... \sum_{x_n} P(X_1 = x_1, X_2 = x_2, ... X_n = x_n) = 1
$$

## Цепи Маркова

**Цепь Маркова** — это последовательность событий, где каждое следующее состояние зависит только от текущего, а не от всей истории. Это называется марковским свойством (или «отсутствием памяти»)

**Марковское свойство** - вероятность оказаться в $j$ на следующем шаге зависит только от текущего $i$, а не от всей предыдущей истории:
$$
P(X_{t+1} = j | X_{t} = i, X_{t-1}, ..., X_{0}) = P(X_{t+1} = j | X_{t} = i)
$$

**Матрица переходов**
| | С | О | Д |
|-| - | - | - |
|С|0,7|0,2|0,1|
|О|0,3|0,4|0,3|
|Д|0,2|0,3|0,5|

Где:
- С - Солнечно
- О - Облачно
- Д - Дождливо

Пример:если сегодня Солнечно, то по первой строке в таблице завтра вероятность погоды: {С:0,7 О:0,2 Д:0,1}

Свойства матрицы переходов:
- Все элементы неотрицательны: $P_{i,j} \ge 0$
- Сумма каждой строки = 1: $\sum_j P_{i,j} = 1$

**Как работать с цепью:**

Допустим, сегодня Солнечно, тогда: $\pi_0 = (1,0,0)$

Завтра: умножаем вектор на матрицу:
$$
\pi_1 = \pi_0 * P = (1,0,0) * \begin{bmatrix}
0.7 & 0.2 & 0.1 \\
0.3 & 0.4 & 0.3 \\
0.2 & 0.3 & 0.5
\end{bmatrix} = (0.7, 0.2, 0.1)
$$

Послезавтра:
$$
\pi_2 = \pi_1 * P = (0.7, 0.2, 0.1) * P = (0.57, 0.25, 0.18)
$$

По итогу общая формула:
$$
\pi_n = \pi_0 * P^n
$$
*Если что, тут всё логично. Подставляя вместо $\pi_1$ результат выше, у нас как раз получится умножение матрицы на матрицу*

**Стационарное распределение** - если продолжать умножение, распределение сходится к некоторому вектору $\pi$, который не меняется (со сути это собственный вектор матрицы с собственным значением 1)
$$
\pi = \pi * P
$$
Причём степень матрицы никак уже не повлияет на этот вектор:
$$
\pi = \pi * P^n
$$

Этот вектор будет показывать общую тенденцию (например, для погоды: (С:46% О:28% Д:26%)) и к нему ты неизбежно придёшь спустя множества количества итераций (хотя там есть исключеня)

*В целом у цепей Маркова есть различные виды и классификации состояний (по типу есть поглощающее состояние, где цепь заканчивается в определённой позиции). Подробнее посмотреть по надобности*
