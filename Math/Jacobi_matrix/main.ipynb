# Матрица Якоби

Общий вид:
$$
J = \begin{pmatrix}
\dfrac{\partial f_1}{\partial x_1} & \dfrac{\partial f_1}{\partial x_2} & \cdots & \dfrac{\partial f_1}{\partial x_n} \\[10pt]
\dfrac{\partial f_2}{\partial x_1} & \dfrac{\partial f_2}{\partial x_2} & \cdots & \dfrac{\partial f_2}{\partial x_n} \\[10pt]
\vdots & \vdots & \ddots & \vdots \\[6pt]
\dfrac{\partial f_m}{\partial x_1} & \dfrac{\partial f_m}{\partial x_2} & \cdots & \dfrac{\partial f_m}{\partial x_n}
\end{pmatrix}
$$

Это буквально тот же самый градиент, но тут этих градиентов несколько и от разных функций  

Например, у нас есть 3 функции:  
$$
f(x,y,z) = x^2y + z
$$
$$
f(x,y,z) = 5x + sin(y) + 2z
$$
$$
f(x,y,z) = xyz
$$
Соотвествующая матрица Якоби для этих функций будет выглядеть так:
$$
J = \begin{array}{c|ccc}
 & \partial x & \partial y & \partial z \\ \hline
\partial f_1 & 2xy & x^2 & 1 \\
\partial f_2 & 5 & \cos(y) & 2 \\
\partial f_3 & yz & xz & xy
\end{array}
$$

То есть тут каждая ячейка это частная производная соотвествующей функции и переменной
