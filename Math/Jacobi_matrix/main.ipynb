# Матрица Якоби

Это буквально тот же самый градиент, но тут этих градиентов несколько и от разных функций  

Например, у нас есть 3 функции:  
$$
f(x,y,z) = x^2y + z
$$
$$
f(x,y,z) = 5x + sin(y) + 2z
$$
$$
f(x,y,z) = xyz
$$
Соотвествующая матрица Якоби для этих функций будет выглядеть так:
$$
J = \begin{array}{c|ccc}
 & \partial x & \partial y & \partial z \\ \hline
\partial f_1 & 2xy & x^2 & 1 \\
\partial f_2 & 5 & \cos(y) & 2 \\
\partial f_3 & yz & xz & xy
\end{array}
$$

То есть тут каждая ячейка это частная производная соотвествующей функции и переменной
