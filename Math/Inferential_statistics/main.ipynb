# Инференциальная статистика

## Популяция и выборка

**Популяция (генеральная совокупность)** - это все объекты, которые нас интересуют (все люди в мире; все детали, произведённые на заводе; все деревья в лесу). $N$ - размер популяции

**Выборка** - это часть популяции, которую мы реально наблюдаем (1000 случайных людей; 50 деталей с конвейера; 200 деревьев). $n$ - размер выборки

**Параметр и статистика** - основной прикол в том, что мы не можем знать средний рост всех людей в мире $\mu$, но зато мы можем посчитать среднюю для выборки $\bar{x}$. По итогу первое это параметр, он описывает популяцию, а второе это статистика, она описывает выборку

**Репрезентативность** - выборка должна представлять популяцию. То есть похожа на неё по всем важным характеристикам  
Пример хорошей выборки: случайные 1000 человек из разных регионов, возрастов, профессий  
Пример плохой выборки: 1000 человек, опрошенных только в фитнес-клубе. Процент курящих в спортзале не равен проценту по всей стране  
Если выборка нерепрезентативна — выводы неверны. Это главная проблема статистики  

### Виды выборок

- **Простая случайная** - каждый объект популяции имеет равный шанс попасть в выборку
- **Систематическая** - берём каждый $k$-й объект (10-ый, 20-ый, 30-ый и т.д.)
- **Стратифицированная** - делим популяцию на группы (страты) и из каждой берём пропорционально (из каждой группы поравну)
- **Кластерная** - делим популяцию на кластеры (например, города), выбираем несколько кластеров и исследуем всех внутри (в отличии от прошлой, тут каждый выбранный кластер мы исследуем ПОЛНОСТЬЮ)
- **Удобная (convenience)** - берём то, что под рукой

### Смещение выборки

Выборка систематически отличается от популяции, примеры:
- Смещение выжившего: изучаем успешные компании — не видим провалившихся
- Смещение самоотбора: опрос на сайте — отвечают только те, кому интересно
- Смещение удобства: опрос студентов днём — работающие не попадают
- Смещение неответа: часть людей не отвечает — их мнение теряется

Последствия: оценки смещены. Среднее по выборке не равно среднему по популяции  
Как бороться: случайный отбор, стратификация, контроль смещений

**Точечная оценка** - это одно число, которое мы предлагаем в качестве оценки параметра

| Параметр | Точечная оценка | Формула |
|-|-|-|
| Среднее $\mu$ | $\bar{x}$ | $\frac{1}{n} \sum x_i$ |
| Дисперсия $\sigma^{2}$ | $s^{2}$ | $\frac{1}{n-1}\sum(x_i - \bar{x})^2$ |
| Доля $p$ | $\hat{p}$ | $\frac{\text{число успехов}}{n}$ |

## Свойства хороших оценок

*Это не те формулы что нужно считать, а просто доказательства для оценок* 

**Несмещённость (unbiasedness)** - в среднем оценка совпадает с параметром:
$$
E[\hat{\theta}] = \theta
$$
где:
$\hat{\theta}$ - оценка
$\theta$ - истинный параметр

**Состоятельность (consistency)** - чем больше выборка, тем ближе оценка к параметру
$$
\hat{\theta}_n \to \theta \text{ при } n \to \infty
$$

**Эффективность (efficiency)** - среди несмещённых оценок — та, у которой меньше дисперсия  

### Стандартная ошибка

**Стандартная ошибка (SE)** - это **стандартное отклонение** оценки или насколько сильно оценка колеблется от выборки к выборке (или ещё проще (на примере средней), допустим получили число $SE$, тогда $\mu$ лежит в диапазоне [$\bar{x} - SE; \bar{x} + SE$])

Формула для среднего:
$$
SE(\bar{x}) = \frac{\sigma}{\sqrt{n}}
$$
где: 
- $\sigma$ - стандартное отклонение популяции (обычно оно нам неизвестно)
- $n$ - количество элементов в выборке

Альтернатива если неизвестна $\sigma$:
$$
SE(\bar{x}) \approx \frac{s}{\sqrt{n}}
$$
где s - стандартное отклонение выборки

*Важно! Это формула лишь для средней величины. Для каждой оценки своя формула. К примеру для доли стандартная ошибка будет:*
$$
SE(\hat{p}) = \sqrt{\frac{\hat{p}(1 - \hat{p})}{n}}
$$
*А для дисперсии она сложнее и зависит от распределения*

### Интервальная оценка

**Интервальная оценка** - это диапазон, в котором с высокой вероятностью лежит параметр

То есть: 
$$
\text{Доверительный интервал } = \bar{x} \pm z * SE
$$
Где $z$ подбирается по таблице: 
| Уровень доверия |  $z$  |
|        -        |   -   |
|       68%       | 1.00  |
|       90%       | 1.645 |
|       95%       | 1.96  |
|       99%       | 2.576 |

*Если повторить эксперимент много раз (взять много выборок) и каждый раз строить интервал, то 95% (или любой другой уровень доверия) этих интервалов накроют истинное значение популяции*

## Проверка гипотез

**Проверка гипотез (Hypothesis Testing)** - это метод проверить утверждение о всей популяции на основе выборки

Например, друг говорит: «Я на вкус отличаю Coca-Cola от Pepsi»  
Мы подозреваем, что он врёт (или просто угадывает)

Формулируем гипотезы:
- $H_0$ - он не отличает. Просто угадывает
- $H_1$ - он действительно отличает

Сама проверка (мы использовуем тут метод "биномиальный тест". Есть и другие, если нужно будет, изучим позже):  
Мы даём ему 10 стаканов (по 5 каждого бренда, порядок случайный). Он пробует и говорит, где что  
**Результат**: он угадал 9 из 10

Вопрос: «Если он просто угадывает $(H_0)$, насколько вероятно угадать 9 из 10?»   
Считаем (биномиальное распределение, $p = 0.5$):
- Вероятность угадать $9 = \binom{10}{9} * 0.5^{10} \approx 0.0098$
- Вероятность угадать $10 = \binom{10}{10} * 0.5^{10} \approx 0.001$
- Сумма = P-value $\approx 0.0107$ (примерно 1%)

* $p < 0.05$ (наш случай) - «Если бы он просто угадывал, такое случилось бы только в 1% случаев. Слишком странно для $H_0$. Отвергаю $H_0$. Скорее всего, он действительно отличает (либо ему очень сильно повезло)»
* $p \ge 0.05$ - «Такое могло случиться случайно. Не отвергаю $H_0$. Не могу утверждать, что он отличает»

**Ключевые моменты (запомнить навсегда):**
1. $H_0$ - всегда «скучный» сценарий: «эффекта нет», «разницы нет», «всё обычно»
2. $H_1$ - то, что мы пытаемся доказать: «эффект есть», «разница есть»
3. p-value — это вероятность увидеть такие данные, ЕСЛИ $H_0$ верна. Не вероятность $H_0$!
4. $p < 0.05$ $\to$ данные слишком странные для $H_0$ $\to$ отвергаем $H_0$
5. $p \ge 0.05$ $\to$ данные нормальные для $H_0$ $\to$ не отвергаем $H_0$
6. Мы никогда не доказываем $H_1$. Мы только отвергаем или не отвергаем $H_0$

- **Ошибка I рода ($\alpha$)** - "Ложная тревога" (то самое значение 0.05 при сравнении с p-value, то есть мы можем это значение менять)
- **Ошибка II рода ($\beta$)** - "Пропущенная цель" (зависит от $n$ - чем больше, тем меньше ошибка. И от $\alpha$ - при её росте падает и наоборот)
- **Мощность** - вероятность правильно отвергнуть $H_0$ (обычно стремятся к $\ge 0.80$)


| | \( H_0 \) верна | \( H_0 \) неверна |
|---|---|---|
| Не отвергли | ✓ | ✗ Ошибка II (\( $\beta$ \)) |
| Отвергли | ✗ Ошибка I (\( $\alpha$ \)) | ✓ Мощность (\( $1-\beta$ \)) |

**Как увеличить мощность:**
- увеличить ($n$)
- уменьшить ($\sigma$ - стандартное отклонение)
- увеличить ($\alpha$) (но это ↑ ошибка I)
- увеличить эффект (почти как $n$, но тут мы не объекты буквально добавляем, а как бы изменяем условия, чтобы реальный эффект проявился сильнее)

## Тесты

| Тест | Когда применять | Пример |
|---|---|---|
| **Z-тест** | Среднее vs число, **σ известна** | Редко на практике |
| **t-тест** | Среднее vs число, **σ неизвестна** | Средний рост = 170? |
| **χ²-тест** | **Категориальные** данные | Связан ли пол с курением? |
| **ANOVA** | **3+ группы**, сравнить средние | Баллы в 3 классах |
| **Манн–Уитни** | **2 группы**, данные **не нормальные** | Удовлетворённость клиентов |
| **Шапиро–Уилк** | **Проверка нормальности** (малая выборка) | Можно ли использовать t-тест? |
| **Колмогоров–Смирнов** | **Проверка распределения** (большая выборка) | Соответствует ли N(0,1)? |

**1. Проверить среднее vs число.**
- σ известна → **Z-тест**
- σ неизвестна → **t-тест**

**2. Сравнить две группы.**
- Нормальные данные → **t-тест** (двухвыборочный)
- Не нормальные → **Манн–Уитни**

**3. Сравнить 3+ группы.**
- Нормальные → **ANOVA**
- Не нормальные → **Крускал–Уоллис**

**4. Связь категорий.**
- → **χ²-тест**

**5. Проверить нормальность.**
- Малая выборка (n < 50) → **Шапиро–Уилк**
- Большая выборка → **Колмогоров–Смирнов**

## Закон больших чисел (ЗБЧ)

Чем больше наблюдений — тем ближе среднее к истинному значению
$$
\bar{x}_n \to \mu \text{ при } n \to \infty
$$

## Центральная предельная теорема (ЦПТ)

Сумма (или среднее) многих независимых случайных величин распределена нормально - независимо от того, как распределены исходные величины

Пример с кубиком. Кубик — равномерное распределение (не нормальное) (1/6, 1/6, 1/6, 1/6, 1/6, 1/6). Берём среднее 30 бросков. Повторяем 1000 раз. Распределение этих средних будет нормальным (в виде горы или колокола)
