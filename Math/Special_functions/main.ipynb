# Специальные функции

## Гамма-функция Г($\alpha$)

$\Gamma(\alpha)$ - это обобщение факториала. Если $\alpha$ целое число, то формула принимает вид:
$$
n! = \Gamma(n + 1)
$$
$$
\Gamma(z) = (z - 1)!
$$
Для других случаев:
$$
\Gamma(z) = \int_0^\infty t^{z-1} e^{-t} \, dt, \quad \operatorname{Re}(z) > 0
$$

*$\operatorname{Re}(z)$ - означает, что если число комплексное, то достаточно чтобы положительным была только первая часть числа без $i$. То есть действительная часть*
