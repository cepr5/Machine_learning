# Отношения

Допустим, есть множество людей:  

$A = \{Аня, Боря, Вася\}$  

Отношение «быть знакомым» связывает тех, кто знаком: 
- Аня знакома с Борей
- Боря знаком с Васей
- Аня не знакома с Васей

(Это лишь одна из вариантов отношения, отношения между элементами могут быть любыми. К примеру "<","=", "брат" и т.д.)

---

Отношение $R$ между отношениями $A$ и $B$ - это подмножество их декартова произведения $A \times B$  
Пример:  
$A = \{1,2,3\}$  
$B = \{a,b\}$  
$A \times B = \{(1,a), (1,b), (2,a), (2,b), (3,a), (3,b)\}$  
Отсюда допустим отношение ($R$) есть только у: \{(1,a), (1,b), (3,b)\}. Что доказывает: $R \subseteq A \times B$  

**Заметка:** ещё обозначение отношения между элементами можно писать так $aRb$  

## Свойства бинарных отношений

У некоторого отношения R могут быть следующие свойства:  

**Рефлексивность** - каждый элемент связан сам с собой. То есть когда есть два одинаковых множества ($R \subseteq A \times A$), то $(a_1,a_1) \in R $  
Пример: A = (1,2,3), R = "=", $(1=1, 2=2, 3=3) \in R$  

**Симметричность** - если $(a, b) \in R$, то и $(b, a) \in R$. То есть связь работает в обе стороны  
Пример: Если Петя знаком с Андреем, то и Андрей знаком с Петей  

**Антисимметричность** - если $(a, b) \in R$ и $(b, a) \in R$, то a = b. То есть если связь работает в обе стороны, то это один и тот же элемент  
Пример: a <= b и b <= a, то a == b  

**Транзитивность** - если $(a, b) \in R$ и $(b, с) \in R$, то $(a, с) \in R$. То есть через цепочку можно перейти напрямую  
Пример: если 1 < 2 и 2 < 3, то и 1 < 3

## Отношение эквивалентности (разбиение на чистые классы)

Отношение эквивалентности — это отношение, которое одновременно:
- Рефлексивно
- Симметрично
- Транзитивно

Отношение эквивалентности разбивает множество на группы, где внутри группы все элементы «одинаковы», а между группами — нет. То есть позволяет делать чистые классы  
Пример:  отношение "быть одноклассником"  
- Каждый сам себе одноклассник
- Если Аня однаклассник Бори, то Боря одноклассник Ани
- Если Аня и Боря одноклассник и Боря и Вася одноклассник, то и Аня и Вася тоже одноклассник  


Логично, что это даёт разбиение на различные классы обучающихся

## Отношение порядка

Отношение порядка — это отношение, которое упорядочивает элементы  

Так же как и отношение эквивалентности оно зависит от *свойств бинарных отношений*. Однако эти свойства разные для каждого из четырёх типов, которые составляются из:
- Нестрогий порядок (<=)
- Строгий порядок (<)
* Частичный порядок (не все элементы сравнимы)
* Линейный (полный) порядок (все элементы сравнимы)

(Более подробно про 4 типа можно посмотреть когда понадобится)

## Функции как отношения

Функция — это частный случай отношения

Свойства функции:
- Инъекция: разные входы → разные выходы
- Сюръекция: каждый элемент B достигается
- Биекция — это и инъекция, и сюръекция одновременно
