# Диагонализация

Диагонализация — это, по сути, попытка представить сложное действие матрицы как простую операцию растяжения вдоль главных осей.  

**Условие для диагонализации**: 
1. У матрицы A должно быть n линейно независимых собственных векторов (где n — это размер матрицы, например, для матрицы 3×3 нужно 3 вектора).
2. Матрица A должна быть квадратной.

$A = P * D * P^{-1}$, где:  
**A** - исходная матрица,  
**P** и **P^{-1}** - это матрица, составленная из собственных векторов матрицы A,  
**D** - это диагональная матрица, на диагонали которой стоят собственные значения матрицы A.  

Если это прдеставить в виде действий, то получится так, справо налево: (Переход на новый базис - Растяжения - Возврат в исходный базис)  

Формула может быть полезна когда например хотим возвести матрицу в степень:  
$A^{n} = P * D^{n} * P^{-1}$

Как найти разложение в коде:

```python
import numpy as np

A = np.array([[4, 1],
              [1, 3]])

print(f"Исходная матрица A: \n{A}")

eigenvalues, P = np.linalg.eig(A) 
print(f"Массив собственных значений: \n{eigenvalues}")
print(f"Матрица, где каждый столбец - это собственный вектор: \n{P}")

D = np.diag(eigenvalues)
print(f"Диагональная матрица D из собственных значений: \n{D}")

# Обратная матрица P
P_inv = np.linalg.inv(P)

A_reconstructed = P @ D @ P_inv
print(f"Изначальная матрица: \n{A_reconstructed}")
```
