# Матрица Гессе

Общий вид:
$$
H = \begin{pmatrix}
\dfrac{\partial^2 f}{\partial x_1^2} & \dfrac{\partial^2 f}{\partial x_1 \partial x_2} & \cdots & \dfrac{\partial^2 f}{\partial x_1 \partial x_n} \\[10pt]
\dfrac{\partial^2 f}{\partial x_2 \partial x_1} & \dfrac{\partial^2 f}{\partial x_2^2} & \cdots & \dfrac{\partial^2 f}{\partial x_2 \partial x_n} \\[10pt]
\vdots & \vdots & \ddots & \vdots \\[6pt]
\dfrac{\partial^2 f}{\partial x_n \partial x_1} & \dfrac{\partial^2 f}{\partial x_n \partial x_2} & \cdots & \dfrac{\partial^2 f}{\partial x_n^2}
\end{pmatrix}
$$

Мы берём одну функцию и для каждой отдельной переменной по два раза находим частную производную. Лучше всего это понять на примере:  
Функция:  
$$
f(x,y) = x^2 + 3y^2
$$
Первые частные производные:  
$$
\frac{\partial f}{\partial x} = 2x
$$
$$
\frac{\partial f}{\partial y} = 6y
$$
Вторые частные производные:  
$$
\frac{\partial f}{\partial x^2} = 2
$$
$$
\frac{\partial f}{\partial y^2} = 6
$$
$$
\frac{\partial f}{\partial x\partial y} = 0
$$
$$
\frac{\partial f}{\partial y\partial x} = 0
$$
Итоговая матрица будет выглядеть так:  
$$
H = \begin{pmatrix}
2 & 0 \\
0 & 6
\end{pmatrix}
$$
