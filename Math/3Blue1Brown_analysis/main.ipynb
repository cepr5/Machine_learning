## Производная

Подходя к пониманию производной, рассмотрим пример с вычислением скорости в какие-то моменты времени. Мы не можем узнать скорость в точке, но можем узнать её в каком-то коротком промежутке dt:  
![Картинка](Derivative2.PNG)  

Основная сложность производной в том, что её dt стремится к нулю. То есть это не какое-то малое значение и не сам 0. Некий парадокс. Однако благодаря этому мы получаем что-то вроде касательной к графику в этой точке. Или ещё можно назвать как изменение в одно мгновение:  
![Картинка](Derivative3.PNG)  

Мы можем попробовать посчитать изменение в графике $t^3$ взяв небольшой промежуток. Берём формулу что была выше, подставляем значения, раскрываем скобки и упрощаем. Получится та картина что видим ниже. Но учитывая то, что dt стремится к нулю, мы можем отбросить правую часть решения и тем самым получим уже знакомую нам производную, а именно $3*t^2$:  
![Картинка](Derivative4.PNG)

## Сложение, умножение и композиция(функция в функции) производных

Сложение производных или (правило дифференцирования суммы) - $\frac{d(sin(x) + x^2)}{dx}$ == $\frac{d(sin(x))}{dx} + \frac{d(x^2)}{dx}$
Данное равенство объясняется рисунком ниже. Если простым языком, то мы делая небольшой шаг вперёд влюбом случаи складываем получившиеся значения прироста:  
![Картинка](Derivative10.PNG)  

Умножение внутри производной. Мы можем расматривать это как нахождение площади между двух функций. Где в зависимости от x меняется площадь. И мы хотим узнать как изменится площадь в точке x + dx. Как и ранее, мы просто складываем получившиеся прямоугольники, при этом принебрегая $dx^2$. Отсюда и получаем формулу умножения производных или (правило дифференцирования произведения) - $\frac{d(sin(x) * x^2)}{dx}$ == $\frac{d(sin(x))}{dx} * x^2 + sin(x) * \frac{d(x^2)}{dx}$ :
![Картинка](Derivative11.PNG)  

Если мы умножим на константу, то в производной эта константа так и останется. То есть $\frac{d(2*sin(x))}{dx} = 2*cos(x)$ :  
![Картинка](Derivative12.PNG)  

Композиция производных или (дифференцирование сложной функции) - $\frac{d(sin(x^2))}{dx} == \frac{d(sin(f))}{df} * \frac{d(x^2)}{dx}$, где $f = x^2$.  
Дабы избежать путаницы, вот итоговый ответ: $cos(x^2) * 2x$. То есть подставляя f мы делаем вид, что она просто x и решаем как обычно, после подставляя то, что под ней было.  
Объяснение на картинке не очень понял (а именно почему мы в конечном счёте умножаем):  
![Картинка](Derivative13.PNG)  

Пытаясь вычислить значение производной $2^x$ за 1 промежуток времени, можем придти к выводу, что увеличение равняется самому числу в моменте времени. Иными словами, в точке x=3, где результат будет 8, наше увеличение будет: $2^{x+1} - 2^x = 2^4 - 2^3 = 16 - 8 = 8$. Как видно прирост совпал со значением в точке. Важно запомнить одну важную деталь! Нахождение любой производной это отношение маленького шага $dy$ от маленького шага $dx$ или же ($\frac{dy}{dx}$) и прикол в том, что даже если $y$ вырожен не явно, это правило всё равно остаётся таким же. Уравнение показанное на картинке называется неявная кривая:  
![Картинка](Derivative19.PNG)  

Для нахождения производной в подобных ситуациях используют "Неявное дифференцирование":  
![Картинка](Implicit_differentiation.PNG)  

Что бы его понять, взглянем на так называемое "Взаимосвязные величины" в задаче с лестницей. Тут мы допустим хотим узнать, как изменится x(t), где t - время. Мы можем разобрать нашу формулу теоремы Пифагора как показано на рисунке и как и раньше просто взять производную от правой части:  
![Картинка](Ladder.PNG)  

Но есть другой вариант решения и тут важно понять одну важную вещь. Вот допустим у нас есть какая-то функция y(t), мы без понятия как она работает. Для того, что бы взять от неё производную по t мы просто говорим, что мол это будет равно $\frac{dy}{dt}$ и просто оставляем это, где y = y(t). И в нашем примере, мы видим, что эта загадочная $y$ изменяется на -1. И это и есть производная от этой нашей загадочной функции. То есть тут прикол в том, что мы оставляем решение неизвестным до тех пор, пока у нас не появятся данные, что бы вычислить. Ещё можно сказать, что по итогу мы узнаём насколько изменится x(t), если y(t) изменяется на -1. Ну и справа ноль, так-как мы берём производную с двух сторон:
