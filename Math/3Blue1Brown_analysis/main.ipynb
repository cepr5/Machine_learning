Формулы для понимания объяснений:  
- Площадь круга: $\pi*r^2$  
- Длина окружности: $2 * \pi * r $ (Под длиной понимается условно так: если бы окружность была планетой, какое расстояние мне нужно пройти вокруг неё, что бы дойти до того места где я уже был?)
- Площадь прямоугольного треугольника: $0.5 * a * b$ (Где a и b катеты этого прямоугольника)

dr (от латинского differentia — разность) - это некоторая очень малая величина. Обозначающая очень малую разницу между одним значением и следующим. То есть мы делаем микроскопический шаг и значение этого шага и есть dr:  
![Картинка](Dr.PNG)

Исходя из картинки выше, мы можем выстроить каждое кольцо круга на графике, где провести линию длины окружности и каждая полоска будет краюшком слегка её касаться. Если мы будем всё больше уменьшать dr, то всё точнее будем приближаться к вычислению нашей площади круга, но уже через треугольник:

Применять подобные формулировки можно в разных ситуациях, где мы хотим складывать какие-то величины. К примеру, мы можем найти какой путь проехала машина сложив все её скорости в малых промежутках dt:  
![Картинка](Any_task.PNG)  

Интеграл - это функция, которая позволяет найти площадь под каким-то графиком:  
![Картинка](Integral.PNG)

Но если функция не такая удачная и у нас не получается треугольник, то что бы найти площадь, нам нужно разбить фигуру на очень маленькие прямоугольники. Где:
- f(A) - значение функции в этой точке
- dx - очень малый промежуток
- dA - площадь итогового прямоугольника

Немного переставив значения в площади треугольника, мы получим формулу производной, где dx стремится к нулю:  
![Картинка](Derivative.PNG)  

Подходя к пониманию производной, рассмотрим пример с вычислением скорости в какие-то моменты времени. Мы не можем узнать скорость в точке, но можем узнать её в каком-то коротком промежутке dt:  
![Картинка](Derivative2.PNG)  

Основная сложность производной в том, что её dt стремится к нулю. То есть это не какое-то малое значение и не сам 0. Некий парадокс. Однако благодаря этому мы получаем что-то вроде касательной к графику в этой точке. Или ещё можно назвать как изменение в одно мгновение:  
![Картинка](Derivative3.PNG)
