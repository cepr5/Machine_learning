## Сложение, умножение и композиция(функция в функции) производных

Сложение производных или (правило дифференцирования суммы) - $\frac{d(sin(x) + x^2)}{dx}$ == $\frac{d(sin(x))}{dx} + \frac{d(x^2)}{dx}$
Данное равенство объясняется рисунком ниже. Если простым языком, то мы делая небольшой шаг вперёд влюбом случаи складываем получившиеся значения прироста:

Умножение внутри производной. Мы можем расматривать это как нахождение площади между двух функций. Где в зависимости от x меняется площадь. И мы хотим узнать как изменится площадь в точке x + dx. Как и ранее, мы просто складываем получившиеся прямоугольники, при этом принебрегая $dx^2$. Отсюда и получаем формулу умножения производных или (правило дифференцирования произведения) - $\frac{d(sin(x) * x^2)}{dx}$ == $\frac{d(sin(x))}{dx} * x^2 + sin(x) * \frac{d(x^2)}{dx}$ :
![Картинка](Derivative11.PNG)  

Если мы умножим на константу, то в производной эта константа так и останется. То есть $\frac{d(2*sin(x))}{dx} = 2*cos(x)$ :  
![Картинка](Derivative12.PNG)

## Неявное дифференцирование

Важно запомнить одну важную деталь! Нахождение любой производной это отношение маленького шага $dy$ от маленького шага $dx$ или же ($\frac{dy}{dx}$) и прикол в том, что даже если $y$ вырожен не явно, это правило всё равно остаётся таким же. Уравнение показанное на картинке называется неявная кривая:  
![Картинка](Derivative19.PNG)  

Для нахождения производной в подобных ситуациях используют "Неявное дифференцирование":  
![Картинка](Implicit_differentiation.PNG)  

Что бы его понять, взглянем на так называемое "Взаимосвязные величины" в задаче с лестницей. Тут мы допустим хотим узнать, как изменится x(t), где t - время. Мы можем разобрать нашу формулу теоремы Пифагора как показано на рисунке и как и раньше просто взять производную от правой части:  
![Картинка](Ladder.PNG)  

Но есть другой вариант решения и тут важно понять одну важную вещь. Вот допустим у нас есть какая-то функция y(t), мы без понятия как она работает. Для того, что бы взять от неё производную по t мы просто говорим, что мол это будет равно $\frac{dy}{dt}$ и просто оставляем это, где y = y(t). И в нашем примере, мы видим, что эта загадочная $y$ изменяется на -1. И это и есть производная от этой нашей загадочной функции. То есть тут прикол в том, что мы оставляем решение неизвестным до тех пор, пока у нас не появятся данные, что бы вычислить. Ещё можно сказать, что по итогу мы узнаём насколько изменится x(t), если y(t) изменяется на -1. Ну и справа ноль, так-как мы берём производную с двух сторон:  
![Картинка](dy_dx.PNG)  
![Картинка](dy_dx2.PNG)  

Важный момент для понимания производной! Любая производная, это условно просто скорость. Мы просто задаёмся вопросом насколько изменится величина $y$ если совсем чучуть изменить $x$. То есть в задаче с лестницой, говоря что $\frac{dy}{dt} = -1$ мы говорим, что за какой-то малый промежуток $t$ наша $y$ станет -1. И собственно точно так же, за какой-то малый промежуток времени наша $x$ станет $\frac{4}{3}$. 

Ещё момент. Почему мы подставляем 1 м/с. Да потому что если скорость постоянна (не меняется со временем), то: $1 м/с = \frac{0,1 м}{0,1 с} = \frac{0,01 м}{0,01 с} = ...$   
То есть скорость будет равна, как бы мы её не показали. Именно поэтому наша 1 секунда может встать в роли dt

Неявное дифференцирование решается двумя способами. Либо мы всё делим на $dx$, тем самым узнавая что получится если пройти малый шаг по $x$, либо берём загадочную $dt$(допустим время) и находим обе части от $dt$, после $dt$ сокращаем умножив обе части на него и получаем нужное нам уравнение ($dt$ просто помогает найти решение для неявного диференцирования, поэтому он нам потом не нужен). Оба уравнения дают один и тот же результат, но в них подставляются разные значения по итогу:  
Само уравнение: $x^2 + y^2 = 5^2$ (формула окружности в радиусом 5)  
Неявное дифференцирование:  
1. $2x + 2y\frac{dy}{dx} = 0$ - по одной dx (Производная по x)(и не забываем о производной от сложной функции, где $y^2$ это сложная функция)  
2. $2x*dx + 2y*dy = 0$ - полный дифференциал  
Разбор варианта 2:  
$\frac{(x^2 + y^2)}{dt} = \frac{5^2}{dt}$  
$\frac{x^2}{dt} + \frac{y^2}{dt} = 0$  
$2x\frac{dx}{dt} + 2y\frac{dy}{dt} = 0$  
$(2x\frac{dx}{dt} + 2y\frac{dy}{dt}) * dt = 0 * dt$  
$2x\frac{dx}{dt} * dt + 2y\frac{dy}{dt} * dt = 0$  
$2x*dx + 2y*dy = 0$  

Ещё один нюанс неявного дифференцирования. Если мы возьмём точку за пределами окружности, то по получившейся формуле неявного дифференцирования всё равно получим ответ. Но этот ответ будет бессмысленнен для нашей окружности. Исходя из этого, мы можем к примеру взять любой $x$, подставить его в уравнение нашей окружности и узнать $y$ и только после смотреть производную. Или например взять свои $x$ и $y$, но всё равно проверить их по изначальной формуле и только после смотреть производную

### Средняя высота графика

Для этого необходимо интеграл поделить на промежуток интеграла

1. Что бы найти среднюю, мы должны каждую возможную высоту графика (а этих высот бессконечно много) разделить на их количество, то есть тоже можно сказать бессконечность
2. Однако если всё же представить конечное число точек, допустим шириной 0.1. То общее количество всех интервалов равно $\pi / 0.1$
3. Если обобщить это на бессконечное количество возможных малых широт, то можем записать это как $\pi / dx$
4. Следуя базовым алгебраическим операциям мы можем $dx$ вынести наверх
5. И наверху получится ничто иное как интеграл. Где мы как раз находим все прямоугольнички умножением всех высот и ширин

## Ряды Тейлора

Основная суть рядов в том, что бы сделать другую функцию, которая будет максимально приближенна к исходной. Благодаря этому с ней становится проще работать (вычислять, диференцировать, интегрировать и т.д.). Однако важно помнить, что приближение работает только вблизи точки где мы его делаем. Если далеко уйти от точки, приближение перестанет повторять изначальную функцию  

1. Сначала ищем $c_0$. В точке $x=0$ функция выдаст $1$. А первый элемент как раз отвечает за высоту, остальное сокращается подставлением нуля. И получается, что $c_0 = 1$  
2. Далее смотрим на $c_1$. Производная функции всё так же в точке $x=0$ даст $0$. Мы так же берём производную и в нашем многочлене, и подставляем получившийся $0$ к $c_1$
3. Смотрим на $c_2$. Находим от нашей функции уже квадратичную производную и получаем $-1$. И тут важный нюанс! Так же от нашего многочлена берём квадратичную производную и видим, что выходит $2c_2x$. А у нас ранее получилось $-1$. Что бы всё сошлось, наша $c_2$ становится равной $-\frac{1}{2}$. И т.д. моожно до бессконечности

![Картинка](Taylors_Rows.PNG)  
![Картинка](Taylors_Rows2.PNG)  
![Картинка](Taylors_Rows3.PNG)

### Другой способ визуализации производных

"Насколько функция чувствительно к небольшим изменениям аргумента(x)" это ещё один способ понимания производной:  
![Картинка](Derivative25.PNG)  

Ещё способ визуализации, это построить одну прямую где точками отобразить значения $x$, а на второй точками отобразить значения $y$. И таким образом смотря на плотность точек допустим для функции $x^2$ в точке $1$ на прямой отображающей $y$ мы увидим, что точки растягиваются примерно в $2$ раза. То есть при более мелком масштабе это буквально выглядит как умножение на $2$. И производная и будет равняться этому значению в этой точке. Логично, что где-то будет сжиматься, например, в точке $1/4$ точки буду сжиматься в $1/2$ раза. В точке $0$ вообще все точки буквально сожмутся в $0$:
