# There is/are

is - в единственном числе  
are - во множественном

**Утверждение:**
| There is/are | Предмет | Где находится |
|-|-|-|
| There is | a sofa | in my room |
| There are | books | on the table |

**Вопрос:**
| Is/Are there | Предмет | Где находится | Ответ на вопрос |
|-|-|-|-|
| Is there | a car | in the street? | Yes, there is/No, there isn't |
| Are there | photos | in the book? | Yes, there are/No, there aren't |

**Отрицание:**
| There isn't/aren't | Предмет | Где находится |
|-|-|-|
| There isn't | a sofa | in my room |
| There aren't | books | on the table |
